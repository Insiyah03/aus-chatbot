```python
import json
from collections import Counter

with open("data/processed/courses.json", "r", encoding="utf-8") as f:
    courses = json.load(f)

print("Total courses:", len(courses))

print("\nCourses per page:")
page_counts = Counter(c["page"] for c in courses)

for page, count in sorted(page_counts.items()):
    print(f"Page {page}: {count}")

print("\nLast 10 courses:")
for course in courses[-10:]:
    print(
        f"{course['course_code']} | "
        f"{course['title']} | "
        f"page {course['page']}"
    )
```

```python
print("\nPotentially suspicious courses:")

for c in courses:
    if not c["title"] or len(c["title"]) < 3:
        print(c)

    if len(c["text"]) < 50:
        print("SHORT:", c)
```

```python
import json

with open("data/processed/courses.json", encoding="utf-8") as f:
    courses = json.load(f)

print("Courses outside pages 36-37 and 187-248:\n")

for course in courses:
    page = course["page"]

    if page not in [36, 37] and not (187 <= page <= 248):
        print(
            course["course_code"],
            "|",
            course["title"],
            "| page",
            page
        )
```
