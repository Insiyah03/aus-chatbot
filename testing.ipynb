```python
import json
from collections import Counter

with open("data/processed/courses.json", "r", encoding="utf-8") as f:
    courses = json.load(f)

print("Total courses:", len(courses))

print("\nCourses per page:")
page_counts = Counter(c["page"] for c in courses)

for page, count in sorted(page_counts.items()):
    print(f"Page {page}: {count}")

print("\nLast 10 courses:")
for course in courses[-10:]:
    print(
        f"{course['course_code']} | "
        f"{course['title']} | "
        f"page {course['page']}"
    )
```

```python
import pymupdf

PDF_PATH = r"C:\Users\insiy\Documents\Projects_Extracurricular\AUS_RAG_Assistant\data\aus_ug_catalog_25-26.pdf"  # adjust if needed


def inspect_page(page_number):
    doc = pymupdf.open(PDF_PATH)

    page = doc[page_number - 1]

    print(f"Page: {page_number}")
    print(f"Page height: {page.rect.height}")
    print("=" * 80)

    blocks = page.get_text("blocks")

    for i, block in enumerate(blocks):

        if len(block) < 5:
            continue

        x0, y0, x1, y1, text = block[:5]

        text = text.strip()

        if not text:
            continue

        print(f"\nBlock {i}")
        print(f"y0: {y0:.2f}")
        print(f"y1: {y1:.2f}")
        print(f"text: {text}")

    doc.close()
```

```python
inspect_page(169)
```
